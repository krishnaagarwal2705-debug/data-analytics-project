# Customer Orders Analysis

This notebook uses Python lists, tuples, dictionaries, loops, and sets to classify customer purchases and generate business insights.

```python
from collections import Counter, defaultdict

# List of customer names
customers = [
    'Ava Patel', 'Ben Lewis', 'Carlos Ruiz', 'Divya Shah',
    'Ethan Cole', 'Farah Khan', 'Grace Wong', 'Henry Brooks'
]

# Each tuple is: (customer, product, price, category)
orders = [
    ('Ava Patel', 'Laptop', 850, 'Electronics'),
    ('Ava Patel', 'T-Shirt', 25, 'Clothing'),
    ('Ben Lewis', 'Headphones', 80, 'Electronics'),
    ('Ben Lewis', 'Jeans', 60, 'Clothing'),
    ('Carlos Ruiz', 'Blender', 45, 'Home Essentials'),
    ('Divya Shah', 'Smartphone', 600, 'Electronics'),
    ('Divya Shah', 'Hoodie', 50, 'Clothing'),
    ('Divya Shah', 'Lamp', 35, 'Home Essentials'),
    ('Ethan Cole', 'Cookware Set', 70, 'Home Essentials'),
    ('Ethan Cole', 'Towel Set', 20, 'Home Essentials'),
    ('Farah Khan', 'Dress', 75, 'Clothing'),
    ('Farah Khan', 'Headphones', 45, 'Electronics'),
    ('Grace Wong', 'Pillow', 30, 'Home Essentials'),
    ('Grace Wong', 'Mouse', 25, 'Electronics'),
    ('Henry Brooks', 'TV', 400, 'Electronics'),
    ('Henry Brooks', 'Sofa Cover', 90, 'Home Essentials'),
    ('Henry Brooks', 'T-Shirt', 20, 'Clothing')
]
```

## Store and organize customer orders

```python
# Dictionary: customer name -> list of ordered products
customer_orders = {customer: [] for customer in customers}
for customer, product, price, category in orders:
    customer_orders[customer].append(product)

# Dictionary: product -> category
product_categories = {product: category for _, product, _, category in orders}
available_categories = set(product_categories.values())

print('Available categories:', ', '.join(sorted(available_categories)))
print('Ava Patel orders:', customer_orders['Ava Patel'])
```

## Classify customers and calculate category sales

```python
def classify_customer(total_spend):
    if total_spend > 100:
        return 'High-value buyer'
    if total_spend >= 50:
        return 'Moderate buyer'
    return 'Low-value buyer'

total_spend = {customer: 0 for customer in customers}
category_revenue = defaultdict(float)
customer_categories = {customer: set() for customer in customers}
product_frequency = Counter()

# Loop through orders to calculate the required summaries
for customer, product, price, category in orders:
    total_spend[customer] += price
    category_revenue[category] += price
    customer_categories[customer].add(category)
    product_frequency[product] += 1

classifications = {customer: classify_customer(total) for customer, total in total_spend.items()}

for customer in customers:
    print(f'{customer}: ${total_spend[customer]:,.0f} - {classifications[customer]}')
```

## Use set operations and generate insights

```python
high_value_customers = [customer for customer in customers if classifications[customer] == 'High-value buyer']
electronics_customers = {customer for customer, categories in customer_categories.items() if 'Electronics' in categories}
clothing_customers = {customer for customer, categories in customer_categories.items() if 'Clothing' in categories}
multiple_category_customers = {customer for customer, categories in customer_categories.items() if len(categories) > 1}
common_electronics_and_clothing = electronics_customers & clothing_customers
electronics_only = electronics_customers - clothing_customers
top_customers = sorted(total_spend.items(), key=lambda item: item[1], reverse=True)[:3]
most_frequent_products = [product for product, count in product_frequency.items() if count == max(product_frequency.values())]

print('High-value customers:', ', '.join(high_value_customers))
print('Most frequently purchased products:', ', '.join(sorted(most_frequent_products)))
print('Customers buying from multiple categories:', ', '.join(sorted(multiple_category_customers)))
print('Customers buying both Electronics and Clothing:', ', '.join(sorted(common_electronics_and_clothing)))
print('Electronics customers without Clothing purchases:', ', '.join(sorted(electronics_only)))
```

## Final report

```python
print('CUSTOMER ORDERS ANALYSIS REPORT')
print('=' * 34)
print('\nSales by category')
for category, revenue in sorted(category_revenue.items(), key=lambda item: item[1], reverse=True):
    print(f'- {category}: ${revenue:,.0f}')

print('\nTop three customers')
for rank, (customer, spend) in enumerate(top_customers, start=1):
    print(f'{rank}. {customer}: ${spend:,.0f}')

highest_category, highest_revenue = max(category_revenue.items(), key=lambda item: item[1])
print('\nKey business insight')
print(f'{highest_category} is the strongest category, generating ${highest_revenue:,.0f} in sales.')
```
